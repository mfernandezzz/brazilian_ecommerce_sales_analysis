```python
products = pd.read_csv('/content/olist_products_dataset.csv')
products.columns
```

```python
products = products[['product_id', 'product_category_name', 'product_name_lenght', 'product_weight_g', 'product_length_cm', 'product_height_cm', 'product_width_cm']]
products.columns
```

```python
products.head()
```

```python
products.dtypes
```

```python
products.isnull().sum()
```

```python
products.fillna(0, inplace=True)
products.isnull().sum()
```

```python
# change the columns with dtype float to int
float_cols  = ['product_name_lenght', 'product_weight_g', 'product_length_cm', 'product_height_cm', 'product_width_cm']
products[float_cols] = products[float_cols].astype('int64')
products.dtypes
```

```python
products.to_csv('products.csv', index=False)
```
