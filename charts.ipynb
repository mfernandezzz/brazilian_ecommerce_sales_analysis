```python
import pandas as pd
```

```python
revenue_day_2017 = pd.read_csv('/content/revenue_day_2017.csv')
revenue_day_2017.head()
```

```python
revenue_day_2017['purchase_date'] = pd.to_datetime(revenue_day_2017['purchase_date'])
revenue_day_2017.info()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(15, 7))
sns.lineplot(x='purchase_date', y='day_revenue', data=revenue_day_2017)
plt.title('Daily Revenue in 2017')
plt.xlabel('Purchase Date')
plt.ylabel('Daily Revenue')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
payment_type_orders = pd.read_csv('/content/payment_type_orders.csv')
payment_type_orders.head()
```

```python
import plotly.express as px

# Create the donut chart
fig = px.pie(
    payment_type_orders,
    names='payment_type',
    values='orders',
    title='Proportion of Orders by Payment Type',
    hole=0.4 # This creates the donut shape
)

# Display the donut chart
fig.show()
```

```python
import matplotlib.pyplot as plt

# Get data for the pie chart
labels = payment_type_orders['payment_type']
sizes = payment_type_orders['orders']

# Create a figure and a set of subplots
fig, ax = plt.subplots(figsize=(8, 8))

# Create the pie chart
ax.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=45)

# Equal aspect ratio ensures that pie is drawn as a circle.
ax.axis('equal')

plt.title('Proportion of Orders by Payment Type')
plt.show()
```

```python
payment_type_revenue = pd.read_csv('/content/payment_type_revenue.csv')
payment_type_revenue.head()
```

```python
import matplotlib.pyplot as plt

# Get data for the pie chart
labels = payment_type_revenue['payment_type']
sizes = payment_type_revenue['revenue']

# Create a figure and a set of subplots
fig, ax = plt.subplots(figsize=(8, 8))

# Create the pie chart
ax.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=45)

# Equal aspect ratio ensures that pie is drawn as a circle.
ax.axis('equal')

plt.title('Proportion of Revenue by Payment Type')
plt.show()
```

```python
year_month_revenue = pd.read_csv('/content/year_month_revenue.csv')
year_month_revenue.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set the ggplot style
plt.style.use('ggplot')

plt.figure(figsize=(14, 7))
sns.lineplot(data=year_month_revenue, x='month', y='revenue', hue='year', marker='o')
plt.title('Monthly Revenue per Year')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.xticks(range(1, 13), ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(title='Year')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set the ggplot style
plt.style.use('ggplot')

plt.figure(figsize=(16, 8))
# Create a bar plot using seaborn
sns.barplot(data=year_month_revenue, x='month', y='revenue', hue='year', palette='viridis')
plt.title('Monthly Revenue per Year')
plt.xlabel('Month')
plt.ylabel('Revenue')
# Set x-axis ticks to display month names
plt.xticks(ticks=range(0, 12), labels=['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend(title='Year', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
payment_type_products_revenue = pd.read_csv('/content/top_five_products_revenue_payment_type.csv')
payment_type_products_revenue.head()
```

```python
# drop column position
payment_type_products_revenue.drop(columns=['position'], inplace=True)
payment_type_products_revenue.head()
```

```python
# rename column payments to payment
payment_type_products_revenue.rename(columns={'payments': 'payment'}, inplace=True)
payment_type_products_revenue.head()
```

```python
import plotly.express as px

# Create the treemap
fig = px.treemap(
    payment_type_products_revenue,
    path=['payment', 'product'],
    values='revenue',
    title='Product Revenue Proportion by Payment Type'
)

# Display the treemap
fig.show()
```
